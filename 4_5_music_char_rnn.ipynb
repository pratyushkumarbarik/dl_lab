```python
import numpy as np
import tensorflow as tf 
```

```python
with open("D:\DL\datasets\datasets\music.abc","r",encoding="utf-8") as f:
    text = f.read()
text = text[:20000]
print(text[:500])
print(len(text))
```

```python
chars = sorted(set(text))

char_to_idx = {c:i for i,c in enumerate(chars)}
idx_to_char = {i:c for i,c in enumerate(chars)}

encoded = np.array([char_to_idx[c] for c in text])
```

```python
seq_length = 100

X = []
y = []

for i in range(len(encoded)-seq_length):
    X.append(encoded[i:i+seq_length])
    y.append(encoded[i+seq_length])

X = np.array(X)
y = np.array(y)

print(X.shape, y.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(len(chars), 64),
    tf.keras.layers.SimpleRNN(128),
    tf.keras.layers.Dense(len(chars), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy"
)

model.summary()
```

```python
model.fit(X, y, epochs=3, batch_size=64)
```

```python
seed = X[0].tolist()
generated = []

for i in range(500):
    pred = model.predict(np.array([seed]), verbose=0)
    next_char = np.argmax(pred[0])

    generated.append(idx_to_char[next_char])
    seed = seed[1:] + [next_char]

music = "".join(generated)

print(music)
```

```python
abc_music = "X:1\nT:Generated Music\nM:4/4\nK:C\n" + music

with open("generated_music.abc", "w") as f:
    f.write(abc_music)

print("ABC music saved!")
```
