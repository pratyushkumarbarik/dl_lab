```python
import pandas as pd
import nltk
import re 

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
```

```python
df = pd.read_csv("D:\DL\datasets\datasets\quora.csv")
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
df = df.sample(30,random_state=42).reset_index(drop=True)
df = df.dropna(subset=["question1","question2"])
df.shape
```

```python
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
df['tokens1'] = df["question1"].apply(word_tokenize)
df['tokens2'] = df["question2"].apply(word_tokenize)
```

```python
def remove_stopwords(tokens):
    return [word for word in tokens if word.lower() not in  stop_words]
df['token1_clean'] = df['tokens1'].apply(remove_stopwords)
df['token2_clean'] = df['tokens2'].apply(remove_stopwords)

df[['token1_clean','token2_clean']].head()
```

```python
def stemming(tokens):
    return [stemmer.stem(word) for word in tokens]
df['stem1'] = df['token1_clean'].apply(stemming)
df['stem2'] = df['token2_clean'].apply(stemming)
df[['stem1','stem2']].head()
 
```

```python
def lemma(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

df['lemma1'] = df['token1_clean'].apply(lemma)
df['lemma2'] = df['token2_clean'].apply(lemma)

df[['lemma1','lemma2']].head()
```
